```python
from dataclasses import dataclass

@dataclass
class Corrente:
  """
  Vazão: vazão da corrente em kg/h
  Fração: fração mássica em kg/kg
  """
  vazao: float
  fracao: float
```

```python
def misturador(c1: Corrente, c2: Corrente):
  """
  Calcula a vazão e a fração mássica de uma mistura de correntes.
  Entradas:
    c1: Corrente
    c2: Corrente
  Saída:
    F: vazão total
    x: fração mássica total
  """
  # Primeira corrente
  Fa = c1.vazao
  xa = c1.fracao

  # Segunda corrente
  Fb = c2.vazao
  xb = c2.fracao

  # vazão total
  F = Fa + Fb

  # fração mássica total
  x = (Fa*xa + Fb*xb)/F

  return F, x
```
